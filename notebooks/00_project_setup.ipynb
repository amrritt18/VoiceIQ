```python
from datasets import load_dataset, Audio

nonblind = load_dataset(
    "urgent-challenge/urgent2025-sqa",
    split="nonblind_test",
    streaming=True
)

nonblind = nonblind.cast_column(
    "audio",
    Audio(decode=False)
)

sample_nonblind = next(iter(nonblind))

print(sample_nonblind)
```

```python
for i, s in enumerate(nonblind):
    print(i, s["sample_id"], "MOS =", s["mos"])

    if i >= 9:
        break
```

```python
blind_mos = load_dataset(
    "urgent-challenge/urgent2025-sqa",
    split="blind_test_mos",
    streaming=True
)

blind_mos = blind_mos.cast_column(
    "audio",
    Audio(decode=False)
)

sample_blind_mos = next(iter(blind_mos))

print(sample_blind_mos)
```

```python
for i, s in enumerate(blind_mos):
    print(i, s["sample_id"], "MOS =", s["mos"])

    if i >= 9:
        break
```

```python
print(blind_mos)
```

```python
print(blind_mos.features)
```

```python
mos_value = []

for i,s in enumerate(blind_mos):
    if s['mos'] is None:
        continue
    mos_value.append(s['mos'])

    if i >= 99:
        break
print('Labled sample inspected:', len(mos_value))
print('Firts 10 mos value:', mos_value[:10])
```
